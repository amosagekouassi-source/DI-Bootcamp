# Daily Challenge: Precision Prompting for Control & Output Quality

This notebook follows the prompt development cycle to transform complex policy text into clear, employee-facing content.

### Scenario
You work on the Learning & Communications Team. Your task is to turn a dense internal policy update into a concise microlearning snippet.

```python
# Original Policy Text
policy_text = """
Employees must ensure that all remote access to internal systems is established via the approved secure VPN. \
Under no circumstances should unsecured connections or personal devices lacking endpoint protection be used \
to access proprietary data or sensitive communications.
"""
print("Policy Text Loaded.")
```

## Step 1: Craft a Prompt That Controls Output Tone, Format, and Length

**Goal:** Rewrite the policy in a friendly tone, using bullet points, paraphrasing (no direct quotes), and staying under 75 words.

```python
# Step 1: Crafting the initial prompt
# Ensuring policy_text is defined in case the previous cell wasn't run
policy_text = """
Employees must ensure that all remote access to internal systems is established via the approved secure VPN. \
Under no circumstances should unsecured connections or personal devices lacking endpoint protection be used \
to access proprietary data or sensitive communications.
"""

prompt_step_1 = f"""
You are a helpful member of the Learning & Communications Team.

Task: Rewrite the input text provided below into a friendly and clear snippet for our internal newsletter.

Constraints:
1. Use a warm, professional, and clear tone.
2. Organize the information using bullet points.
3. Paraphrase the content entirely; do not use direct quotes from the source.
4. The total output must be under 75 words.

Input Text:
{policy_text}
"""

print("Step 1 Prompt is ready. You can copy the text below for your AI model:")
print("----------------------------------------------------------")
print(prompt_step_1)
```

## Step 2: Evaluate the Output

| Criteria | Guiding Questions |
| :--- | :--- |
| **Relevance** | Does it stick to the source? |
| **Clarity** | Is it easier for non-technical staff? |
| **Structure** | Are there clear bullet points? |
| **Tone** | Is it friendly/internal? |
| **Length** | Under 75 words? |
| **Accuracy** | Any missing security details? |

## Step 3: Detect and Mitigate Hallucinations

If the model added information not in the original text, use this section to refine your prompt with strict constraints.

```python
# Step 3: Revised Prompt to Mitigate Hallucinations
# Use this if the AI added new technologies or rules not in the source text.

prompt_step_3 = f"""
You are a helpful member of the Learning & Communications Team.

Task: Rewrite the input text provided below into a friendly snippet.

Strict Constraints:
1. Only use information provided in the input text.
2. Do NOT add new recommendations, technologies, or policy terms.
3. Use bullet points.
4. Keep it under 75 words.

Input Text:
{policy_text}
"""

print(prompt_step_3)
```

## Step 4: Paraphrasing Deep Dive (Junior Intern Audience)

**Goal:** Plain language, max 4 bullet points, supportive tone, no jargon.

```python
# Step 4: Junior Intern Audience Prompt

prompt_step_4 = f"""
You are a supportive mentor explaining security to a junior intern.

Task: Paraphrase the policy text using plain language and short phrases.

Constraints:
1. Use no more than 4 bullet points.
2. Avoid all corporate or legal jargon.
3. Tone: Supportive and informative.

Input Text:
{policy_text}
"""

print(prompt_step_4)
```
