## 1. Understanding LLM Evaluation

**Explain why evaluating LLMs is more complex than traditional software:**
Traditional software is deterministic (same input always equals same output) and has clear pass/fail criteria. LLMs are probabilistic, meaning they can generate different creative responses to the same prompt. Their 'correctness' is often subjective, context-dependent, and covers a vast open-ended space of possible outputs, making fixed test cases insufficient.

**Identify key reasons for evaluating an LLM’s safety:**
Safety evaluation is critical to prevent the model from generating toxic, biased, or harmful content. It ensures the model doesn't leak private data (PII), provide instructions for illegal acts, or reinforce social stereotypes that could cause real-world harm.

**Describe how adversarial testing contributes to LLM improvement:**
Adversarial testing (red-teaming) involves intentionally trying to break the model or trick it into violating its guardrails. By identifying these 'edge cases' or vulnerabilities, developers can use techniques like RLHF (Reinforcement Learning from Human Feedback) or fine-tuning to make the model more robust and safer.

**Discuss the limitations of automated evaluation metrics and how they compare to human evaluation:**
Automated metrics (like BLEU/ROUGE) focus on word overlap rather than meaning or factual truth. They cannot judge nuance, tone, or creativity. Human evaluation is the 'gold standard' because humans understand context and intent, but it is slow, expensive, and difficult to scale compared to instant automated scores.

## 2. Applying BLEU and ROUGE Metrics

We will use the `evaluate` library (or `nltk`/`rouge_score`) to calculate these metrics.

```python
# Install necessary libraries
!pip install evaluate rouge_score sacrebleu
```

```python
import evaluate

bleu = evaluate.load("bleu")
rouge = evaluate.load("rouge")

# BLEU Exercise
ref_bleu = ["Despite the increasing reliance on artificial intelligence in various industries, human oversight remains essential to ensure ethical and effective implementation."]
gen_bleu = "Although AI is being used more in industries, human supervision is still necessary for ethical and effective application."

results_bleu = bleu.compute(predictions=[gen_bleu], references=[ref_bleu])
print("BLEU Results:", results_bleu)

# ROUGE Exercise
ref_rouge = ["In the face of rapid climate change, global initiatives must focus on reducing carbon emissions and developing sustainable energy sources to mitigate environmental impact."]
gen_rouge = "To counteract climate change, worldwide efforts should aim to lower carbon emissions and enhance renewable energy development."

results_rouge = rouge.compute(predictions=[gen_rouge], references=[ref_rouge])
print("ROUGE Results:", results_rouge)
```

## 3. Perplexity Analysis

**Model A vs Model B:**
Perplexity is a measure of how well a probability distribution predicts a sample. Since Model A assigns a higher probability (0.8) to the correct word than Model B (0.4), **Model A has lower perplexity**. In NLP, a lower perplexity score indicates that the model is more confident and accurate in its predictions.

**Perplexity of 100:**
A perplexity of 100 means that on average, the model is as confused as if it had to choose between 100 equally likely words at each step. For a large vocabulary, this might be acceptable, but for simple tasks, it indicates poor performance. To improve it, we could increase the training data, fine-tune on domain-specific text, or adjust the model architecture.

## 6. Comparative Analysis of Evaluation Methods

**Chosen NLP Task:** Text Summarization

**Comparison of Metrics:**
1. **ROUGE:** Measures n-gram overlap between the generated summary and a reference summary. It is excellent for checking recall (how much of the original information is kept) but fails to account for synonyms or grammatical quality.
2. **BERTScore:** Uses contextual embeddings to measure semantic similarity between tokens. Unlike ROUGE, it can recognize that two sentences mean the same thing even if they use different words.
3. **Human Evaluation:** Involves human raters scoring the summary on coherence, conciseness, and factual consistency. It is the most accurate for quality but impossible to automate for large-scale real-time testing.

**Most Appropriate Metric:**
For **Text Summarization**, a combination of **ROUGE-L** and **BERTScore** is currently the industry standard. ROUGE-L ensures that the longest common sequences are preserved (structure), while BERTScore ensures the *meaning* is captured correctly without penalizing the model for creative paraphrasing. However, for a production-ready model, a final round of **Human Evaluation** is essential to check for 'hallucinations' that automated metrics might miss.
