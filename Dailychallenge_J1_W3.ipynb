```python
#Data Preparation

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)  # pour reproductibilité
#Création des lignes
cities = ["Abidjan", "Yamoussoukro", "Daloa", "Korhogo", "Man", "San Pedro",
          "Bouaké", "Bondoukou", "Abengourou", "Danané"]
#Création des colonnes
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

# génération des données
data = np.random.uniform(-5, 35, size=(10, 12))

# création du DataFrame
df = pd.DataFrame(np.round(data, ), index=cities, columns=months)

df
```

```python
#Data Analysis

# a) Température moyenne annuelle par ville
mean_temp = df.mean(axis=1)
print("The average temperature per city :")
print(np.round(mean_temp, ))

# b) Ville la plus chaude et la plus froide
hottest_city = mean_temp.idxmax()
coldest_city = mean_temp.idxmin()

print("The hottest city is:", hottest_city, np.round(mean_temp.max()))
print("The coldest city is:", coldest_city, np.round(mean_temp.min()))
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

df_melted = df.reset_index().melt(id_vars="index")
df_melted.columns = ["City", "Month", "Temperature"]

plt.figure(figsize=(12,6))

sns.lineplot(
    data=df_melted,
    x="Month",
    y="Temperature",
    hue="City",
    marker="o"
)

plt.title("Monthly Temperature Trends for Each City")
plt.xticks(rotation=45)
plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left")

plt.show()
```

```python
plt.figure(figsize=(12,6))
plt.imshow(df, aspect='auto')
plt.colorbar(label="Temperature (°C)")
plt.yticks(range(len(cities)), cities)
plt.xticks(range(len(months)), months, rotation=45)
plt.title("Monthly Temperature Heatmap")
plt.show()
```

## Findings Summary

- Temperatures were randomly generated between -5°C and 35°C.

- Each city has a different annual average.

- The hottest city is Bondoukou.

- The coldest city is Yamoussoukro.

- The heatmap shows seasonal variations and allows you to observe warmer and colder periods.

- The differences between cities are due to random generation.

- The summer months (June–August) generally tend to be warmer.

The boxplot highlights the distribution of temperatures for each city throughout the year.

- Cities with larger boxes exhibit greater temperature variability.
- Cities with smaller boxes have more stable temperatures across the year.
- The median line indicates the central tendency of temperatures.
- No significant outliers were observed in most cities, suggesting a relatively consistent temperature distribution.

Overall, coastal cities tend to show less variation, while inland cities display larger seasonal fluctuations.
