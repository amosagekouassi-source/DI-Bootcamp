```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
from sklearn.decomposition import PCA

Salaries = pd.read_csv('datascience_salaries.csv')
Salaries.head()
```

```python
#Normalization of the 'salary' column
normal_salary = MinMaxScaler()
Salaries['salary'] = normal_salary.fit_transform(Salaries[['salary']])
Salaries.head()
```

```python
Experimented = Salaries.groupby('experience_level').sum()
Experimented
```

```python
Salaries_mean = Salaries.groupby('experience_level')['salary'].mean()
Salaries_mean
```

```python
Salaries_median = Salaries.groupby('experience_level')['salary'].median()
Salaries_median
```

```python
import numpy as np
#Reduction of the number of features in the dataset

# Select only numeric columns for PCA
numeric_salaries = Salaries.select_dtypes(include=np.number)

pca = PCA(n_components=2)
Salaries_pca = pca.fit_transform(numeric_salaries)

# Convert the result back to a DataFrame for easier handling if needed
Salaries = pd.DataFrame(Salaries_pca, columns=[f'PC_{i+1}' for i in range(Salaries_pca.shape[1])])
Salaries.info()
```
