### 1. Load the MNIST dataset
We will use Keras to load the training and testing sets.

```python
import tensorflow as tf
from tensorflow.keras import layers, models, utils
import numpy as np

# Load MNIST dataset
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

print(f"Training data shape: {x_train.shape}, Labels: {y_train.shape}")
print(f"Testing data shape: {x_test.shape}, Labels: {y_test.shape}")
```

### 2. Preprocess data for Fully Connected Neural Network (FCNN)
FCNNs require flattened input (784 pixels) and normalized values.

```python
# Flatten: (60000, 28, 28) -> (60000, 784)
x_train_flat = x_train.reshape(-1, 784).astype('float32') / 255.0
x_test_flat = x_test.reshape(-1, 784).astype('float32') / 255.0

# One-hot encoding labels
y_train_cat = utils.to_categorical(y_train, 10)
y_test_cat = utils.to_categorical(y_test, 10)

print(f"Preprocessed FCNN shape: {x_train_flat.shape}")
```

### 3. Build and train FCNN
A simple model with Dense layers.

```python
fcnn_model = models.Sequential([
    layers.Dense(512, activation='relu', input_shape=(784,)),
    layers.Dense(10, activation='softmax')
])

fcnn_model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

print("Training FCNN...")
fcnn_model.fit(x_train_flat, y_train_cat, epochs=5, batch_size=128, verbose=1)

loss, acc_fcnn = fcnn_model.evaluate(x_test_flat, y_test_cat, verbose=0)
print(f"FCNN Test Accuracy: {acc_fcnn:.4f}")
```

### 4. Preprocess data for Convolutional Neural Network (CNN)
CNNs expect a 4D tensor (samples, height, width, channels).

```python
# Reshape to (28, 28, 1)
x_train_cnn = x_train.reshape(-1, 28, 28, 1).astype('float32') / 255.0
x_test_cnn = x_test.reshape(-1, 28, 28, 1).astype('float32') / 255.0

print(f"Preprocessed CNN shape: {x_train_cnn.shape}")
```

### 5. Build and train CNN
Using Conv2D and MaxPooling2D layers.

```python
import tensorflow as tf
from tensorflow.keras import layers, models, utils

# Ensure data is loaded in case of kernel reset
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

# 1. Preprocess for CNN
x_train_cnn = x_train.reshape(-1, 28, 28, 1).astype("float32") / 255.0
x_test_cnn = x_test.reshape(-1, 28, 28, 1).astype("float32") / 255.0
y_train_cat = utils.to_categorical(y_train, 10)
y_test_cat = utils.to_categorical(y_test, 10)

# 2. Build and train CNN
cnn_model = models.Sequential([
    layers.Input(shape=(28, 28, 1)),
    layers.Conv2D(32, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(64, activation="relu"),
    layers.Dense(10, activation="softmax")
])

cnn_model.compile(optimizer="adam", loss="categorical_crossentropy", metrics=["accuracy"])
print("Training CNN...")
cnn_model.fit(x_train_cnn, y_train_cat, epochs=5, batch_size=128, verbose=1)

_, acc_cnn = cnn_model.evaluate(x_test_cnn, y_test_cat, verbose=0)
print(f"CNN Test Accuracy: {acc_cnn:.4f}")
```

### 6. Compare Performance
CNNs usually outperform FCNNs on image data because they capture spatial hierarchies.

```python
# Final performance comparison
try:
    # Try to evaluate the FCNN if it's in memory
    _, acc_fcnn_final = fcnn_model.evaluate(x_train.reshape(-1, 784)/255.0, utils.to_categorical(y_test, 10), verbose=0)
except NameError:
    # Fallback to the printed accuracy if variables were lost
    print("Note: FCNN model variable not found. Using previous result (0.9798).")
    acc_fcnn_final = 0.9798

print(f"Fully Connected Model Accuracy: {acc_fcnn_final:.4f}")
print(f"Convolutional Neural Network Accuracy: {acc_cnn:.4f}")
print(f"Accuracy Improvement of CNN: {(acc_cnn - acc_fcnn_final)*100:.2f}%")

# Summary: CNNs are superior for image tasks as they preserve spatial hierarchies."
```
