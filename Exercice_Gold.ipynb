```python
#Exercise 2: Comparative Analysis of Retail Data

import glob
import pandas as pd

# 1. BLOC D'IMPORTATION UNIVERSEL
fichiers_trouves = glob.glob("*.csv")
if not fichiers_trouves:
    raise FileNotFoundError(
        "❌ Aucun fichier CSV détecté. Veuillez glisser-déposer votre fichier 'sales data-set.csv' à gauche."
    )

# Chargement automatique de votre fichier sales data-set.csv
chemin_actuel = fichiers_trouves[0]
df = pd.read_csv(chemin_actuel, encoding="ISO-8859-1")

# Nettoyage préventif des noms de colonnes
df.columns = df.columns.str.strip()

print("=== MODE DETECTÉ : DONNÉES STRUCTURÉES (RETAIL SALES) ===")
print(f"Fichier analysé avec succès : '{chemin_actuel}'\n")

# 1. Tendances des ventes globales
total_sales = df['Weekly_Sales'].sum()
print(f"-> Chiffre d'affaires total global : {total_sales:,.2f} $")

# 2. Performance des magasins (Top 3 des meilleurs magasins en ventes cumulées)
top_stores = df.groupby('Store')['Weekly_Sales'].sum().sort_values(ascending=False).head(3)
print("\n-> Top 3 des magasins les plus performants (Chiffre d'affaires) :")
print(top_stores.map(lambda x: f"{x:,.2f} $"))

# 3. Habitudes d'achat des clients (Impact des périodes de fêtes / vacances)
if 'IsHoliday' in df.columns:
    ventes_vacances = df.groupby('IsHoliday')['Weekly_Sales'].mean()
    print("\n-> Ventes hebdomadaires moyennes :")
    print(f"   * Semaines de vacances / fêtes : {ventes_vacances[True]:,.2f} $")
    print(f"   * Semaines normales            : {ventes_vacances[False]:,.2f} $")
```

```python
#Exercise 6: Creating a Synthetic Product Catalog

!pip install faker # Install the faker library
import random
import pandas as pd
from faker import Faker

# 1. Initialisation de Faker
fake = Faker()

# 2. Génération des 500 produits synthétiques
catalogue = []
for i in range(1, 501):
    catalogue.append(
        {
            "ID": f"PROD-{i:03d}",
            "Name": fake.catch_phrase(),
            "Description": fake.paragraph(nb_sentences=2),
            "Price": round(random.uniform(9.99, 999.99), 2),
        }
    )

# 3. Création du DataFrame
catalog_df = pd.DataFrame(catalogue)

print("--- CATALOGUE DE PRODUITS SYNTHÉTIQUES CRÉÉ ---")
print(catalog_df.head(5))
```
