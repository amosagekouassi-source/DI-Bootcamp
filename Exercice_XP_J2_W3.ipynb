```python
#Exercise 1 : Matrix Operations

import numpy as np

# 1. Création d'une matrice 3x3 inversible
matrice = np.array([[2, 1, 3], [0, 5, 6], [1, 0, 2]])

# 2. Calcul du déterminant
det = np.linalg.det(matrice)
print(f"Déterminant : {det:.2f}")

# 3. Calcul de l'inverse (possible car le déterminant n'est pas nul)
if det != 0:
    inverse = np.linalg.inv(matrice)
    print("\nMatrice Inverse :\n", inverse)
else:
    print("\nLa matrice n'est pas inversible (déterminant nul).")
```

```python
#Exercise 6 : Basic Hypothesis Testing

import numpy as np
from scipy import stats  # Module standard pour compléter le test d'hypothèse

np.random.seed(42)
productivity_before = np.random.normal(loc=50, scale=10, size=30)
productivity_after = productivity_before + np.random.normal(
    loc=5, scale=3, size=30
)

# Test T pour échantillons dépendants/appariés (Mêmes employés avant/après)
t_stat, p_value = stats.ttest_rel(productivity_before, productivity_after)

print(f"Statistique T : {t_stat:.4f}")
print(f"P-value       : {p_value}")

if p_value < 0.05:
    print(
        "\nConclusion : Rejet de H0. Le programme de formation a une efficacité statistiquement significative."
    )
else:
    print(
        "\nConclusion : Échec du rejet de H0. Aucune preuve d'efficacité de la formation."
    )
```

```python
#Exercise 8 : Time Series Data Manipulation

import pandas as pd

# 1. Génération de données temporelles journalières pour l'année 2023
dates = pd.date_range(start="2023-01-01", end="2023-12-31", freq="D")
valeurs = range(len(dates))
ts_data = pd.Series(valeurs, index=dates)

# 2. Démonstration du découpage (Slicing) par intervalles de trimestres
print("Janvier à Mars    (T1) : Lignes de", ts_data["2023-01":"2023-03"].shape[0])
print("Avril à Juin      (T2) : Lignes de", ts_data["2023-04":"2023-06"].shape[0])
print("Juillet à Sept.   (T3) : Lignes de", ts_data["2023-07":"2023-09"].shape[0])
print("Octobre à Déc.    (T4) : Lignes de", ts_data["2023-10":"2023-12"].shape[0])
```
