**Exercice 1**: Understanding Data Visualization

1. Why is data visualization important in data analysis?

Reponse :

Data visualization is important because it helps transform complex data into visual formats that are easier to understand. It allows analysts to identify patterns, trends, relationships, and outliers quickly. Visualizations also improve communication of findings and support data-driven decision-making.

2. Purpose of a Line Graph

Reponse :

A line graph is used to display changes or trends over time. It connects data points with lines, making it easy to observe increases, decreases, fluctuations, and overall trends in a dataset.

```python
#Exercice 2: Creating a Line Plot for Temperature Variation

import matplotlib.pyplot as plt

# Temperature data
temperatures = [72, 74, 76, 80, 82, 78, 75]

# Days of the week
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']

# Create line plot
plt.figure(figsize=(8,5))
plt.plot(days, temperatures, marker='o')

# Labels and title
plt.xlabel("Day")
plt.ylabel("Temperature (°F)")
plt.title("Temperature Variation Over a Week")

# Display plot
plt.show()
```

```python
#Exercice 3 : Monthly Sales Bar Chart

import matplotlib.pyplot as plt

# Sales data
sales = [5000, 5500, 6200, 7000, 7500]

# Months
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']

# Create bar chart
plt.figure(figsize=(8,5))
plt.bar(months, sales)

# Labels and title
plt.xlabel("Month")
plt.ylabel("Sales Amount ($)")
plt.title("Monthly Sales")

# Display plot
plt.show()
```

```python
#Exercice 4: Distribution of CGPA

import pandas as pd

df = pd.read_csv("Student_Mental_health.csv")
df.columns
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x='What is your CGPA?',
    color='skyblue'
)

plt.title("Distribution of Students' CGPA")

plt.show()
```

```python
#Exercise 5: Comparing Anxiety Levels Across Different Genders

import seaborn as sns
import matplotlib.pyplot as plt

# Calculate proportions
anxiety_gender = pd.crosstab(
    df['Choose your gender'],
    df['Do you have Anxiety?'],
    normalize='index'
)

anxiety_gender.plot(
    kind='bar',
    stacked=True,
    figsize=(8,5)
)

plt.title("Anxiety Levels Across Genders")
plt.ylabel("Proportion")

plt.show()
```

```python
#Exercice 6 : Age vs Panic Attacks

#Convert Yes/No into numbers
df['PanicNumeric'] = (
    df['Do you have Panic attack?']
    .map({'Yes':1, 'No':0})
)

#Scatter Plot
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x='Age',
    y='PanicNumeric'
)

plt.xlabel("Age")
plt.ylabel("Panic Attack (0 = No, 1 = Yes)")
plt.title("Relationship Between Age and Panic Attacks")

plt.show()
```
