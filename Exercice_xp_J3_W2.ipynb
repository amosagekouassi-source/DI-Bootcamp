```python
#Exercice 2 : Statistiques Descriptives

import numpy as np
from scipy import stats

data = [12, 15, 13, 12, 18, 20, 22, 21]

# Calculs
mean_val = np.mean(data)
median_val = np.median(data)
variance_val = stats.tvar(data)  # Variance de l'échantillon (degré de liberté = 1)
std_dev_val = stats.tstd(data)   # Écart-type de l'échantillon

print(f"Moyenne : {mean_val}")
print(f"Médiane : {median_val}")
print(f"Variance (échantillon) : {variance_val:.2f}")
print(f"Écart-type (échantillon) : {std_dev_val:.2f}")
```

```python
#Exercice 3 : Comprendre les Distributions (Loi Normale)

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Paramètres de la distribution
mean = 50
std_dev = 10

# Générer des points sur l'axe X (de mean - 4*std à mean + 4*std pour bien voir la courbe)
x = np.linspace(mean - 4 * std_dev, mean + 4 * std_dev, 100)

# Calculer la Densité de Probabilité (PDF)
y = norm.pdf(x, mean, std_dev)

# Tracé du graphique
plt.figure(figsize=(8, 4))
plt.plot(x, y, label=f'Loi Normale ($\mu={mean}$, $\sigma={std_dev}$)', color='blue', lw=2)
plt.title('Distribution Normale')
plt.xlabel('Valeurs')
plt.ylabel('Densité de probabilité')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

```python
#Exercice 4 : Application du T-Test (Test d'hypothèse)

import numpy as np
from scipy import stats

# Génération des données
np.random.seed(42) # Pour avoir les mêmes résultats à chaque exécution
data1 = np.random.normal(50, 10, 100)
data2 = np.random.normal(60, 10, 100)

# Exécution du T-test indépendant
t_stat, p_val = stats.ttest_ind(data1, data2)

print(f"Statistique T : {t_stat:.4f}")
print(f"P-value : {p_val:.4e}") # Affichage en notation scientifique car très proche de 0

# Interprétation rapide
if p_val < 0.05:
    print("Interprétation : Les moyennes des deux groupes sont significativement différentes.")
else:
    print("Interprétation : Il n'y a pas de différence significative entre les deux groupes.")
```

```python
#Exercice 5 : Analyse de Régression Linéaire

import numpy as np
from scipy import stats

# Données
house_sizes = np.array([50, 70, 80, 100, 120])
house_prices = np.array([150000, 200000, 210000, 250000, 280000])

# Calcul de la régression
slope, intercept, r_value, p_value, std_err = stats.linregress(house_sizes, house_prices)

print(f"Slope (Pente) : {slope:.2f}")
print(f"Intercept (Ordonnée à l'origine) : {intercept:.2f}")

# 2. Prédire le prix pour 90 m²
size_to_predict = 90
predicted_price = (slope * size_to_predict) + intercept
print(f"Prix prédit pour une maison de 90m² : {predicted_price:,.2f} unités de devise")
```

```python
#Exercice 6 : Comprendre l'ANOVA (Lien avec ton premier message)

from scipy import stats

fertilizer_1 = [5, 6, 7, 6, 5]
fertilizer_2 = [7, 8, 7, 9, 8]
fertilizer_3 = [4, 5, 4, 3, 4]

# Test ANOVA à un facteur
f_stat, p_val = stats.f_oneway(fertilizer_1, fertilizer_2, fertilizer_3)

print(f"F-value : {f_stat:.4f}")
print(f"P-value : {p_val:.5f}")
```

```python
#Exercice 7 : Loi Binomiale (Probability Distributions)

import scipy.stats as stats

# Paramètres de l'exercice
n = 10  # Nombre de lancers de pièce
p = 0.5  # Probabilité d'avoir pile (succès)
k = 5   # Nombre exact de succès voulus

# Calcul de la probabilité avec la loi binomiale (PMF = Probability Mass Function)
probabilite = stats.binomial.pmf(k, n, p) if hasattr(stats, 'binomial') else stats.binom.pmf(k, n, p)

print(f"La probabilité d'obtenir exactement {k} piles est de : {probabilite:.4f}")
print(f"Soit environ : {probabilite * 100:.2f}%")
```

```python
#Exercice 8 : Coefficients de Corrélation

import pandas as pd
from scipy import stats

# 1. Création du jeu de données (DataFrame Pandas)
data = pd.DataFrame({
    'age': [23, 25, 30, 35, 40],
    'income': [35000, 40000, 50000, 60000, 70000]
})

# 2. Calcul du coefficient de Pearson (Relation linéaire)
# On peut utiliser directement la méthode .corr() de Pandas
pearson_corr = data['age'].corr(data['income'], method='pearson')

# 3. Calcul du coefficient de Spearman (Relation monotone / basée sur les rangs)
spearman_corr = data['age'].corr(data['income'], method='spearman')

# 4. Affichage des résultats
print(f"Coefficient de Pearson : {pearson_corr:.2f}")
print(f"Coefficient de Spearman : {spearman_corr:.2f}")

# Interprétation automatique
if pearson_corr == 1.0:
    print("\nInterprétation : Il y a une corrélation positive parfaite (1.0).")
    print("Quand l'âge augmente, le revenu augmente de manière parfaitement linéaire.")
```
