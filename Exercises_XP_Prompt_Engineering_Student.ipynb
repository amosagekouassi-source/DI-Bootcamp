## Exercise 1: Rewrite and Optimize a Vague Prompt
**Scenario**: Manager asks for a LinkedIn post about FlowNest with the vague prompt `"Write something about productivity tips."`

### Tasks
- Identify 3 critical issues with the prompt.
- Rewrite it using: role prompting, clear instructions, and format/tone/length constraints.
- Add a final line to make it reusable for future social posts.

```python
critical_issues = [
    'Lack of specific context: The prompt does not mention the product "FlowNest".',
    'Undefined target audience: It does not specify who the LinkedIn post is for (e.g., CEOs, students).',
    'Missing constraints: There is no guidance on tone, length, or desired call to action.'
]
critical_issues
```

```python
rewritten_prompt = """
Role: You are an expert Social Media Manager specializing in B2B SaaS.
Task: Write a LinkedIn post promoting FlowNest, focusing on productivity tips for remote teams.
Constraints:
- Tone: Professional yet engaging and helpful.
- Format: Start with a hook, followed by 3 bulleted tips, and end with a question to drive engagement.
- Length: Under 200 words.
Reusable Variable: [Insert Topic Here] - use this for future posts about different productivity sub-topics.
"""
rewritten_prompt
```

## Exercise 2: Multi-Part Prompt for Quiz Generation
**Scenario**: Create an educational prompt for 7th graders about volcanic eruptions.

### Tasks
- Ask for: 2-bullet summary + 3 MCQs (1 correct, 2 distractors).
- Require friendly tone, simple vocab (11?13 y/o), and Google Slides-ready output.
- Explain 3 ways your prompt improves on: `Make a quiz for kids about this article.`

```python
quiz_prompt = """
Act as a middle school science teacher.
Topic: Volcanic Eruptions.
Target Audience: 7th graders (11-13 years old).
Instructions:
1. Provide a 2-bullet point summary of how volcanoes erupt using simple vocabulary.
2. Create 3 Multiple Choice Questions (MCQs).
3. Each MCQ must have 1 correct answer and 2 plausible distractors.
4. Tone: Friendly and encouraging.
5. Output Format: Structured for Google Slides (Title, Summary Slide, Quiz Slide).
"""
quiz_prompt
```

```python
improvements = [
    'Persona Adoption: By acting as a teacher, the tone and complexity are automatically adjusted for the 7th-grade level.',
    'Structural Constraints: Defining the exact number of bullets and MCQs prevents the model from being too brief or too wordy.',
    'Format Targeting: Specifically requesting Google Slides-ready output reduces the manual work needed to transport the data.'
]
improvements
```
