2) Discuss: How did the shift to modern NLP impact scalability and efficiency?

- The shift to modern NLP, specifically via Representation Learning, eliminated the bottleneck of manual feature engineering, allowing models to learn features directly from raw data. ## 🌟 Exercise 2: LLM Architecture and Application Scenarios
For each, describe (a) core architectural differences, (b) a real application, (c) why it fits.

### BERT
- **Architecture**: Bidirectional encoder-only; Masked Language Modeling (MLM) and Next Sentence Prediction (NSP).
- **Application**: Search Engine Query Understanding (e.g., Google Search).
- **Why**: Its bidirectional nature allows it to understand the full context of a word based on its surroundings (left and right), which is crucial for disambiguating search intent.

### GPT
- **Architecture**: Decoder-only; autoregressive causal LM (predicts the next token).
- **Application**: Creative Writing Assistant or Code Generation (GitHub Copilot).
- **Why**: The causal masking ensures the model only looks at previous tokens, making it optimized for generating coherent, fluent, and long-form text or code.

### T5
- **Architecture**: Encoder-decoder; text-to-text framework (everything is a string-to-string task).
- **Application**: Document Summarization.
- **Why**: The encoder processes the long input document while the decoder generates a concise summary, benefiting from the specialized roles of both components.

## 🌟 Exercise 3: Benefits and Ethics of Pre-training
- **Benefits**:
  1) **Improved generalization**: Models exposed to diverse web-scale data learn universal language patterns that apply to many domains.
  2) **Less labeled data**: Because the model already 'knows' language, it only needs a few hundred examples to learn a specific downstream task.
  3) **Faster fine-tuning**: Adapting a pre-trained model takes significantly less time than training from scratch.
  4) **Transfer learning**: Knowledge gained from predicting missing words can be 'transferred' to sentiment analysis or translation.
  5) **Robustness**: Massive pre-training helps models handle typos, slang, and varied syntax more gracefully.

- **Ethical concerns**: Pre-training data often contains human biases (gender, race), can generate toxic misinformation, might be used for phishing, and risks leaking private info found in training sets.
- **Mitigations**: Using curated datasets, applying Differential Privacy, implementing safety guardrails (filters), using RLHF (Reinforcement Learning from Human Feedback), and conducting third-party bias audits.

## 🌟 Exercise 4: Transformer Architecture Deep Dive
### Self-Attention & Multi-Head Attention
- **Q/K/V Flow**: Queries (what I'm looking for) are compared against Keys (what I have) to calculate scores. Softmax normalizes these into weights, which are used to compute a weighted sum of Values (the content to keep).
- **Multi-Head Attention**: Multiple 'heads' allow the model to attend to different parts of the sequence simultaneously. It projects the input into different subspaces, capturing diverse linguistic relations.
- **Example sentence**: "The bank of the river was closed by the central bank."
  - **Head 1**: Might focus on the relationship between "bank" and "river" (geographic context).
  - **Head 2**: Might focus on the relationship between "bank" and "central" (financial context).

### Pre-training Objectives
- **MLM vs CLM**: MLM (BERT) masks random tokens to learn context from both sides; CLM (GPT) predicts the next token to learn generation flow.
- **When to prefer**: Use MLM for classification and extraction; use CLM for dialogue, storytelling, and open-ended generation.
- **NSP**: Next Sentence Prediction helped early BERT understand relationships between sentences, but modern models (like RoBERTa) found it unnecessary if training on longer contiguous blocks of text.

### Transformer Model Selection
- **Sentiment on reviews**: Encoder-only (BERT). Justification: Sentiment analysis requires understanding the full context of the review to identify nuance.
- **Conversational chatbot**: Decoder-only (GPT). Justification: Requires high fluency and the ability to generate tokens sequentially.
- **Technical translation**: Encoder-Decoder (T5/BART). Justification: The encoder captures the source language nuances, while the decoder handles the target language structure.

### Positional Encoding
- **Purpose**: Transformers process all tokens at once (parallel), so they lack an inherent sense of word order. Positional encodings add vectors to indicate where each word sits in the sequence.
- **Example failure**: Without positions, the model would treat "The cat ate the mouse" and "The mouse ate the cat" as identical bags of words.

## 🌟 Exercise 6: Softmax Temperature: The Randomness Regulator
### 1) Temperature Scenarios
- **T=0.2**: High confidence, deterministic. The model picks the most likely tokens, resulting in repetitive or 'safe' text.
- **T=1.5**: High entropy, chaotic. The model gives higher weight to unlikely tokens, leading to creative but often nonsensical results.
- **T=1.0**: Neutral. The distribution is unchanged; a balance between logic and variety.

### 2) Application Design
- **Bedtime stories**: T=0.8 to 1.0. High enough for imagination and variety, but low enough to maintain a logical plot.
- **Financial report summaries**: T=0.1 or greedy decoding. Accuracy is paramount; the model must not 'hallucinate' or vary the data.

### 3) Temperature & Bias
- Higher temperature can surface rare biases found in the long tail of the training data. For example, at T=0.2, a model might always complete "The doctor is..." with "he", but at T=1.2, it might alternate between various pronouns but also occasionally output offensive or stereotypical slurs that were suppressed at lower temperatures.

### (Optional) Quick Generation Demo Across Temperatures
Note: Requires `transformers` and model download; skip if offline.


```python
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline
import torch

model_name = "gpt2"
prompt = "The future of artificial intelligence in education is"

pipe = pipeline(
    "text-generation",
    model=AutoModelForCausalLM.from_pretrained(model_name),
    tokenizer=AutoTokenizer.from_pretrained(model_name),
    device=0 if torch.cuda.is_available() else -1,
)

for T in [0.2, 1.0, 1.5]:
    # Using do_sample=True to allow temperature to take effect
    out = pipe(prompt, max_new_tokens=40, temperature=T, do_sample=True, pad_token_id=50256)
    print(f"\n--- Temperature: {T} ---")
    print(out[0]["generated_text"])
```
