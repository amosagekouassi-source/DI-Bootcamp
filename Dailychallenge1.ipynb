Mon document porte sur la production du Cacao en Côte d'Ivoire et s'intitule <<Traçabilité et transparence des chaînes d'approvisionnement du Cacao Côte d'Ivoire et Ghana>> sur https://www.nitidae.org/files/90b65fa3/document_technique_agriculture_redd_marche_ghana_cotedivoire_cacao_tracao.pdf

Ici, le Conseil Café-Cacao a mené un recensement national des producteurs. Les équipes de Data Analysts ont dû traiter un volume de données immense (900.000 ménages de producteurs recensés, plus 1,4 millions d'hectares de parcelles analysés.)
Le rôle de l'analyse de données a été de nettoyer ces données dites brûtes (gérerles valeurs manquantes, éliminer les doublons) et de croiser les informations socio-économiques des planteurs avec des données géographiques. Ensuite, les analystes ont croisé les données de géoréférencement des parcelles avec des images satllites (notamment l'indice de navigation NDVI l'USGS).
Comme résultat, l'analyse de données a permis de voir que avant, l'estimation des surfaces cultivées était approximative. L'analyse de données graphiques a permis de cartographier avec précision la repartion de la production.
Les algorithmes ont prouvé mathématiquement que 60% de la production nationale est désormais concentrée dans la région de l'Ouest tandis que l'Est (la région historique) ne représente plus que 8%.
En croisant l'âge des plantations (donnée quantitative) et l'indice de végétation, les modèles ont expliqué les chutes de productivité: un quart des verges ivoiriens a plus de 30 ans et a dépassé son âge optimal.
Ainsi, grâce à l'analyse des données, on peut maintenant tracer l'origine du Cacao (s'il ne vient pas d'une forêt classée), saisir les fraudeurs (si une coopérative déclare livrer 10 tonnes de Cacao alors que la surface de ses polygones ne permet d'en produire que 3), identifier chaque producteur en lui attribuant un identifiant, suivre le Cacao depuis la brousse jusqu'au Port à partir de l'identifiant du planteur.
